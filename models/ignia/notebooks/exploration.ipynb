```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# paths
img_dir = "../data/cityscapes/leftImg8bit/train/aachen"
mask_dir = "../data/cityscapes/gtFine/train/aachen"

# load one image
img_files = os.listdir(img_dir)
img_path = os.path.join(img_dir, img_files[0])
img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# load corresponding mask
# cityscapes mask files end with _gtFine_labelIds.png
mask_file = img_files[0].replace(
    "_leftImg8bit.png", 
    "_gtFine_labelIds.png"
)
mask_path = os.path.join(mask_dir, mask_file)
mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
```

```python
# visualize
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].imshow(img)
axes[0].set_title(f"Image: {img.shape}")
axes[1].imshow(mask, cmap="tab20")
axes[1].set_title(f"Mask: {mask.shape}, Unique labels: {np.unique(mask)}")
plt.tight_layout()
plt.savefig("exploration.png")
plt.show()
```

```python
print("Image shape:", img.shape)
print("Mask shape:", mask.shape)
print("Unique label IDs in mask:", np.unique(mask))
```
